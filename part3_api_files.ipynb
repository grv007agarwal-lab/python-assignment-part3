## **Task 1 — File Read & Write Basics**

```python
# -------------------------------
# Part A: Write and Append to File
# -------------------------------

file_name = "python_notes.txt"

# Step 1: Write initial content
with open(file_name, "w", encoding="utf-8") as file:
    file.write("Topic 1: Variables store data. Python is dynamically typed.\n")
    file.write("Topic 2: Lists are ordered and mutable.\n")
    file.write("Topic 3: Dictionaries store key-value pairs.\n")
    file.write("Topic 4: Loops automate repetitive tasks.\n")
    file.write("Topic 5: Exception handling prevents crashes.\n")

print("File written successfully.")

# Step 2: Append additional lines
with open(file_name, "a", encoding="utf-8") as file:
    file.write("Topic 6: Functions help reuse code.\n")
    file.write("Topic 7: Modules help organize code.\n")

print("Lines appended.")


# -------------------------------
# Part B: Read and Process File
# -------------------------------

# Step 3: Read file and print numbered lines
lines = []

with open(file_name, "r", encoding="utf-8") as file:
    lines = file.readlines()

print("\n--- File Content ---")
for i, line in enumerate(lines, start=1):
    print(f"{i}. {line.strip()}")

# Step 4: Count total lines
print(f"\nTotal number of lines: {len(lines)}")


# Step 5: Keyword search (case-insensitive)
keyword = input("\nEnter keyword to search: ").strip().lower()

matches = [line.strip() for line in lines if keyword in line.lower()]

print("\n--- Search Results ---")
if matches:
    for line in matches:
        print(line)
else:
    print("No matching lines found.")
```

## **Task 2 — API Integration**

```python
import requests

BASE_URL = "https://dummyjson.com/products"

# -------------------------------
# Step 1: Fetch and Display Products
# -------------------------------
try:
    response = requests.get(f"{BASE_URL}?limit=20")
    response.raise_for_status()  # Raise error for bad status codes
    data = response.json()

    products = data.get("products", [])

    print("ID  | Title                          | Category      | Price     | Rating")
    print("----|--------------------------------|---------------|-----------|-------")

    for p in products:
        print(f"{p['id']:<3} | {p['title'][:30]:<30} | {p['category']:<13} | ${p['price']:<9} | {p['rating']}")

except requests.exceptions.RequestException as e:
    print(f"Error fetching products: {e}")


# -------------------------------
# Step 2: Filter and Sort
# -------------------------------
filtered = [p for p in products if p["rating"] >= 4.5]

# Sort by price descending
filtered_sorted = sorted(filtered, key=lambda x: x["price"], reverse=True)

print("\n--- High Rated Products (≥ 4.5) Sorted by Price ---")
for p in filtered_sorted:
    print(f"{p['title']} | ${p['price']} | Rating: {p['rating']}")


# -------------------------------
# Step 3: Search by Category
# -------------------------------
try:
    response = requests.get(f"{BASE_URL}/category/laptops")
    response.raise_for_status()
    data = response.json()

    laptops = data.get("products", [])

    print("\n--- Laptops Category ---")
    for p in laptops:
        print(f"{p['title']} | ${p['price']}")

except requests.exceptions.RequestException as e:
    print(f"Error fetching laptops: {e}")


# -------------------------------
# Step 4: POST Request (Simulated)
# -------------------------------
new_product = {
    "title": "My Custom Product",
    "price": 999,
    "category": "electronics",
    "description": "A product I created via API"
}

try:
    response = requests.post(f"{BASE_URL}/add", json=new_product)
    response.raise_for_status()

    print("\n--- POST Response ---")
    print(response.json())

except requests.exceptions.RequestException as e:
    print(f"Error in POST request: {e}")
```

## **Task 3 — Exception Handling**

```python
# Function: Safe division with error handling
def safe_divide(a, b):
    try:
        # Step 1: Attempt division
        return a / b

    # Step 2: Handle division by zero
    except ZeroDivisionError:
        return "Error: Cannot divide by zero"

    # Step 3: Handle invalid input types
    except TypeError:
        return "Error: Invalid input types"


# -------------------------------
# Test Cases
# -------------------------------

print(safe_divide(10, 2))     # Expected: 5.0
print(safe_divide(10, 0))     # Expected: Error message
print(safe_divide("ten", 2))  # Expected: Error message


# Function: Safe file reader with error handling
def read_file_safe(filename):
    try:
        # Step 1: Try to open and read file
        with open(filename, "r", encoding="utf-8") as file:
            content = file.read()
            return content

    # Step 2: Handle file not found error
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")

    # Step 3: Always execute this block
    finally:
        print("File operation attempt complete.")


# -------------------------------
# Test Cases
# -------------------------------

# Case 1: File exists
print("\n--- Reading python_notes.txt ---")
content = read_file_safe("python_notes.txt")
if content:
    print(content)

# Case 2: File does not exist
print("\n--- Reading ghost_file.txt ---")
content = read_file_safe("ghost_file.txt")
if content:
    print(content)



import requests

BASE_URL = "https://dummyjson.com/products"

# -------------------------------
# Step 1: Fetch Products (GET)
# -------------------------------
try:
    response = requests.get(f"{BASE_URL}?limit=20", timeout=5)

    # Check HTTP status
    if response.status_code == 200:
        data = response.json()
        products = data.get("products", [])

        print("Products fetched successfully.")
    else:
        print(f"HTTP Error: {response.status_code}")
        products = []

# Handle network-level errors
except requests.exceptions.ConnectionError:
    print("Connection failed. Please check your internet.")
    products = []

except requests.exceptions.Timeout:
    print("Request timed out. Try again later.")
    products = []

except Exception as e:
    print(f"Unexpected error: {e}")
    products = []


# -------------------------------
# Step 2: Fetch Laptops Category
# -------------------------------
try:
    response = requests.get(f"{BASE_URL}/category/laptops", timeout=5)

    if response.status_code == 200:
        laptops = response.json().get("products", [])
        print("\nLaptops fetched successfully.")
    else:
        print(f"HTTP Error: {response.status_code}")
        laptops = []

except requests.exceptions.ConnectionError:
    print("Connection failed. Please check your internet.")
    laptops = []

except requests.exceptions.Timeout:
    print("Request timed out. Try again later.")
    laptops = []

except Exception as e:
    print(f"Unexpected error: {e}")
    laptops = []


# -------------------------------
# Step 3: POST Request
# -------------------------------
new_product = {
    "title": "My Custom Product",
    "price": 999,
    "category": "electronics",
    "description": "A product I created via API"
}

try:
    response = requests.post(f"{BASE_URL}/add", json=new_product, timeout=5)

    if response.status_code in (200, 201):
        print("\nPOST successful:")
        print(response.json())
    else:
        print(f"HTTP Error: {response.status_code}")

except requests.exceptions.ConnectionError:
    print("Connection failed. Please check your internet.")

except requests.exceptions.Timeout:
    print("Request timed out. Try again later.")

except Exception as e:
    print(f"Unexpected error: {e}")



    import requests

BASE_URL = "https://dummyjson.com/products"

while True:
    # Step 1: Take user input
    user_input = input("\n Enter a product ID to look up (1–100), or 'quit' to exit: ").strip()

    # Step 2: Exit condition
    if user_input.lower() == "quit":
        print("Exiting program.")
        break

    # Step 3: Validate input is integer
    if not user_input.isdigit():
        print("Invalid input. Please enter a number between 1 and 100.")
        continue

    product_id = int(user_input)

    # Step 4: Validate range
    if not (1 <= product_id <= 100):
        print("Invalid range. Please enter a number between 1 and 100.")
        continue

    # Step 5: Make API call
    try:
        response = requests.get(f"{BASE_URL}/{product_id}", timeout=5)

        # Step 6: Check status code
        if response.status_code == 200:
            product = response.json()
            print(f"Product: {product['title']} | Price: ${product['price']}")

        elif response.status_code == 404:
            print("Product not found.")

        else:
            print(f"HTTP Error: {response.status_code}")

    # Step 7: Handle exceptions
    except requests.exceptions.ConnectionError:
        print("Connection failed. Please check your internet.")

    except requests.exceptions.Timeout:
        print("Request timed out. Try again later.")

    except Exception as e:
        print(f"Unexpected error: {e}")
```

## **Task 4 — Logging to File**

```python
import requests
from datetime import datetime

LOG_FILE = "error_log.txt"

# -------------------------------
# Helper Function: Write Log
# -------------------------------
def log_error(context, error_type, message):
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    with open(LOG_FILE, "a", encoding="utf-8") as file:
        file.write(f"[{timestamp}] ERROR in {context}: {error_type} — {message}\n")


# -------------------------------
# Step 1: Trigger ConnectionError
# -------------------------------
try:
    requests.get("https://this-host-does-not-exist-xyz.com/api", timeout=5)

except requests.exceptions.ConnectionError as e:
    log_error("fetch_products", "ConnectionError", str(e))


# -------------------------------
# Step 2: Trigger HTTP Error (404)
# -------------------------------
try:
    response = requests.get("https://dummyjson.com/products/999", timeout=5)

    # Manually check for HTTP error
    if response.status_code != 200:
        log_error(
            "lookup_product",
            "HTTPError",
            f"{response.status_code} Not Found for product ID 999"
        )

except requests.exceptions.ConnectionError as e:
    log_error("lookup_product", "ConnectionError", str(e))

except requests.exceptions.Timeout as e:
    log_error("lookup_product", "Timeout", str(e))

except Exception as e:
    log_error("lookup_product", "UnexpectedError", str(e))


# -------------------------------
# Step 3: Read and Print Log File
# -------------------------------
print("\n--- Error Log Contents ---")

with open(LOG_FILE, "r", encoding="utf-8") as file:
    print(file.read())
```
